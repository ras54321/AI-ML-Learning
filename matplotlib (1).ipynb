```python
# Movies Data
# Oscar Winner Movies
oscar_movies = [
    "The Dark Knight", 
    "The Hurt Locker",  
    "The King's Speech", 
    "The Artist",     
    "Argo"
]
oscar_years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]  # in $M
plt.plot(oscar_years,oscar_revenue)
plt.title("Oscar Movies revenue in each year")
plt.xlabel("years")
plt.ylabel("Revenue(in$M)")
```

```python
# Plot multiple
# Non-Oscar Winner Movies
non_oscar_movies = ["Slumdog Millionaire","Avatar","Inception","Hugo","Lincoln"]
non_oscar_years = [2008, 2009, 2010, 2011, 2012]
non_oscar_revenue = [378, 2788, 829, 185, 275]  # in $M
plt.plot(oscar_years,oscar_revenue,color="red",linestyle="--",marker="o",label="oscar movies")
plt.plot(non_oscar_years,non_oscar_revenue,color="green",linestyle="-",marker=",",label="non oscar movies")
plt.title("oscar movies vs non oscar movies revenue")
plt.xlabel("years")
plt.ylabel("Revenue(in$M)")
plt.legend()
```

```python
#bar chart
oscar_movies = [
    "The Dark Knight", 
    "The Hurt Locker",  
    "The King's Speech", 
    "The Artist",     
    "Argo"
]
oscar_years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]  # in $M
plt.bar(oscar_years,oscar_revenue,color="green")
plt.title("revenue in each year for oscar movies")
plt.xlabel("year")
plt.ylabel("revenue(in$M)")
plt.tight_layout()
# Add labels on top of bars
for i in range(len(oscar_revenue)):
    plt.text(oscar_years[i], oscar_revenue[i] + 20, str(oscar_revenue[i]), ha="center")  
    # text(pos of text acc to x-axis, pos of y, text, alignment)
plt.ylim(0, max(oscar_revenue)+200) #lower, upper limit - add space
```

```python
# comparing data with bar charts
oscar_movies = ["The Dark Knight","The Hurt Locker","The King's Speech","The Artist","Argo"]
oscar_revenue = [1005, 170, 427, 133, 232]  # in $M

non_oscar_movies = ["Slumdog Millionaire","Avatar","Inception","Hugo","Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275]  # in $M

years = [2008, 2009, 2010, 2011, 2012]
# Plotting Bar Chart
import numpy as np
x = np.arange(len(years))
width = 0.4

plt.bar(x-width/2, oscar_revenue, width, label="Oscar",color="blue")
plt.bar(x+width/2, non_oscar_revenue, width, label="Non-Oscar",color="#056875")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
plt.title("Yearly revenue of Oscar vs Non-Oscar movies")
plt.legend()
plt.tight_layout()
plt.xticks(x, years) # to show x-axis labels
```

```python
# Scatter Plots
people = ["Person A", "Person B", "Person C", "Person D", "Person E", 
          "Person F", "Person G", "Person H", "Person I", "Person J"]
age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
blood_pressure = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]
colors=["green" if bp<130 else "red" for bp in blood_pressure ]
plt.scatter(age,blood_pressure,cmap="RdBu",c=blood_pressure,marker="^")#color=colors,s=blood_pressure,alpha=0.5,marker="o"
plt.title("Age vs BP")
plt.xlabel("age")
plt.ylabel("BP")
plt.grid()
plt.colorbar(label="BP")
for i in range(len(age)):
    plt.annotate(people[i], xy=(age[i], blood_pressure[i]))
plt.xlim(min(age), max(age)+10)
plt.ylim(min(blood_pressure),max(blood_pressure)+2)
```

```python
# Multiple scatter plots
cities = ["City A", "City B", "City C", "City D", "City E"]

# Winter: Temperature (°C) vs Humidity (%)
winter_temp = [5, 2, 10, 0, 7]
winter_humidity = [80, 75, 65, 85, 70]

# Summer: Temperature (°C) vs Humidity (%)
summer_temp = [25, 30, 28, 35, 27]
summer_humidity = [60, 50, 55, 45, 65]

# Plots
plt.scatter(winter_temp, winter_humidity, color="blue", label="Winter")
plt.scatter(summer_temp, summer_humidity, color="orange", label="Summer")
plt.xlabel("Temperature (°C)")
plt.ylabel("Humidity (%)")
plt.legend()
```

```python
# Pie Charts
expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amounts = [500, 150, 120, 100, 50]
# explode
explode = [0, 0, 0, 0.2, 0]
plt.pie(amounts,labels=expenses,autopct="%1.1f%%",
       wedgeprops={"linewidth":1,
                   "linestyle":"-",
                    "edgecolor": "black"            
       },explode=explode,startangle=0,shadow=True
       )
#plt.style.available
plt.show()
```
