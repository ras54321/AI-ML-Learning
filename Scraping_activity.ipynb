```python
import requests
from bs4 import BeautifulSoup
```

```python
#Fetch Data and store it
page_count = 1
while True:
    URL=f"https://quotes.toscrape.com/page/{page_count}/"
    res=requests.get(URL)
    soup=BeautifulSoup(res.text,"lxml")
    quotes=soup.select("div.quote")
    if not quotes:
        print("No valid page")
        break
    with open(f"scraped data/quotes{page_count}.html","w",encoding="utf-8") as f:
        f.write(res.text)
        print(f"downloaded data from page{page_count}")
    page_count=page_count+1   
     
```

```python
all_quotes = soup.select("div.quote")
life_quotes = []

for q in all_quotes:
    all_tags = []

    for tag in q.select(".tags .tag"):
        all_tags.append(tag.get_text())

    if "life" in all_tags:
        text = q.select_one("span.text").get_text()
        author = q.select_one("small.author").get_text()

        life_quotes.append([text, author])

print(life_quotes)
```

```python
from bs4 import BeautifulSoup

life_quotes = []

# Loop through all 10 saved pages
for page_count in range(1, 11):

    # Open each HTML file
    with open(
        f"scraped data/quotes{page_count}.html",
        "r",
        encoding="utf-8"
    ) as f:
        html_content = f.read()

    soup = BeautifulSoup(html_content, "lxml")

    all_quotes = soup.select("div.quote")

    # Check every quote on this page
    for q in all_quotes:
        all_tags = []

        for tag in q.select(".tags .tag"):
            all_tags.append(tag.get_text())

        # Select only quotes having the "life" tag
        if "life" in all_tags:
            text = q.select_one("span.text").get_text()
            author = q.select_one("small.author").get_text()

            life_quotes.append([text, author])

print(life_quotes)
```
