```python
import requests
URL2="https://www.scrapethissite.com/pages/simple/"
res=requests.get(URL2)
```

```python
if res.status_code==200:
    #print(res.text)
    #print(res.content)
    print(res.headers)
with open("scraped data/data1.html","w",encoding="utf-8") as f:
    f.write(res.text)
```

```python
#working with beautiful soups
from bs4 import BeautifulSoup

with open("scraped data/data1.html", "r", encoding="utf-8") as f:
    html_content = f.read()

soup = BeautifulSoup(html_content, "lxml")
```

```python
#soup.find("h1")
all_h3=soup.find_all("h3")
all_countries=[]
for h3 in all_h3:
    name=h3.get_text(strip=True)
    print(name)
    #print(h3.find_parent("div")["class"])
    population=h3.find_next("div").select("span.country-population")[0].get_text(strip=True)
    print(population)
    all_countries.append([name,population])
    #print(h3.children) DOM
    #print(h3.parent)
    #print(h3.next_sibling)
    
```

```python
import pandas as pd
df=pd.DataFrame(all_countries,columns=["country","population"])
```

```python
df.to_csv("Cleaned Data/data.csv")
```
